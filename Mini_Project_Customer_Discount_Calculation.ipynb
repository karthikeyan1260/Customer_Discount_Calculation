<b> Customer Discount Calculation </b>

```python
for i in range (1,6):
    print("\nCustomer",i)

    Name = input("Enter Customer Name:")
    Amount = float(input("Enter the Amount: Rs."))

    if Amount > 5000:
        Discount = 1000
    elif Amount > 2000:
        Discount = 500
    else:
        Discount = 0

    Final_Amount = Amount - Discount

    print("\nCustomer Name:",Name)
    print("Purchase Amount: Rs.",Amount)
    print("Discount: Rs.",Discount)
    print("Final Payable Amount: Rs.",Final_Amount)
    
```
